リスト2.9 ライブラリのインポート

````python
from langchain.chains import SequentialChain
from langchain.chains import LLMChain
from langchain.prompts import PromptTemplate
import pandas as pd
import matplotlib.pyplot as plt
import json
````

リスト2.10 データ分析プロンプトの定義

````python
# 1. データ分析プロンプト - 顧客データのインサイト抽出
data_analysis_prompt = PromptTemplate(
    input_variables=["customer_data", "analysis_goal"],
    template="""
    あなたはデータアナリストです。以下の顧客データを分析し、{analysis_goal}についての主要なインサイトを3つ抽出してください。
    JSONフォーマットで返してください。


    顧客データ:
    {customer_data}


    出力形式:
    ```json
    {{
        "insights": [
            {{
                "title": "インサイトのタイトル",
                "description": "詳細な説明",
                "business_impact": "ビジネスへの影響"
            }},
            ...
        ]
    }}
    ```
    """
)
````

リスト2.11 提案生成プロンプトの定義

````python
# 2. 提案生成プロンプト - データインサイトからビジネス提案を作成
recommendation_prompt = PromptTemplate(
    input_variables=["insights", "business_type"],
    template="""
    あなたはビジネスコンサルタントです。以下のデータインサイトに基づいて、{business_type}ビジネスのための具体的な改善提案を作成してください。
    各提案は実現可能で具体的であることが重要です。JSONフォーマットで返してください。


    データインサイト:
    {insights}


    出力形式:
    ```json
    {{
        "recommendations": [
            {{
                "title": "提案タイトル",
                "description": "詳細な説明",
                "implementation_steps": ["ステップ1", "ステップ2", ...],
                "expected_outcome": "期待される成果"
            }},
            ...
        ]
    }}
    ```
    """
)
````

リスト2.12 実装計画プロンプトの定義

````python
# 3. 実装計画プロンプト - 提案を実行するための具体的な計画
implementation_prompt = PromptTemplate(
    input_variables=["recommendations"],
    template="""
    あなたはプロジェクトマネージャーです。以下の提案を実装するための30日間の実行計画を作成してください。
    計画には明確なタイムライン、必要なリソース、成功指標を含めてください。
    JSONフォーマットで返してください。


    ビジネス提案:
    {recommendations}


    出力形式:
    ```json
    {{
        "implementation_plan": {{
            "timeline": [
                {{
                    "week": 1,
                    "focus_areas": ["フォーカスエリア1", ...],
                    "key_activities": ["活動1", ...]
                }},
                ...
            ],
            "required_resources": ["リソース1", ...],
            "success_metrics": ["指標1", ...]
        }}
    }}
    ```
    """
)
````

リスト2.13 各プロンプトを使用したLLMChainを作成してSequentialChainで連結

````python
# チェーン1: データ分析
data_analysis_chain = LLMChain(
    llm=chat_model,
    prompt=data_analysis_prompt,
    output_key="insights"
)

# チェーン2: 提案生成
recommendation_chain = LLMChain(
    llm=chat_model,
    prompt=recommendation_prompt,
    output_key="recommendations"
)

# チェーン3: 実装計画
implementation_chain = LLMChain(
    llm=chat_model,
    prompt=implementation_prompt,
    output_key="implementation_plan"
)

# シーケンシャルチェーンの作成
business_advisor_chain = SequentialChain(
    chains=[data_analysis_chain, recommendation_chain, implementation_chain],
    input_variables=["customer_data", "analysis_goal", "business_type"],
    output_variables=["insights", "recommendations", "implementation_plan"],
    verbose=True
)
````

リスト2.14 ビジネスアドバイザーチェーンの実行

````python
# サンプルデータ
customer_data = """
年齢,性別,購入金額,購入頻度(月),最終購入日,好みのカテゴリ
28,女性,15000,2.3,2023-02-15,化粧品
35,男性,8000,1.1,2023-01-20,電子機器
42,女性,25000,3.2,2023-02-28,アパレル
31,男性,5000,0.8,2022-12-05,書籍
45,女性,30000,2.5,2023-02-10,アパレル
29,女性,12000,2.1,2023-01-25,化粧品
38,男性,20000,1.5,2023-02-05,スポーツ用品
52,女性,18000,1.2,2023-01-15,家庭用品
33,男性,7000,0.9,2022-11-30,電子機器
41,女性,22000,2.8,2023-02-20,アパレル
"""

# チェーンを実行
result = business_advisor_chain({
    "customer_data": customer_data,
    "analysis_goal": "購買パターンと顧客セグメント",
    "business_type": "小売"
})
````

リスト2.15 結果を整形して表示

````python
# 結果を整形して表示
def prettify_json(json_str):
    try:
        # JSON文字列から```jsonと```を取り除く
        json_str = json_str.replace("```json", "").replace("```", "").strip()
        data = json.loads(json_str)
        return json.dumps(data, indent=2, ensure_ascii=False)
    except:
        return json_str

# 結果の表示
print("\n===== データ分析のインサイト =====")
print(prettify_json(result["insights"]))
print("\n===== ビジネス提案 =====")
print(prettify_json(result["recommendations"]))
print("\n===== 実装計画 =====")
print(prettify_json(result["implementation_plan"]))
````

````python
# データの可視化
def visualize_results(insights_json, recommendations_json):
    try:
        # JSON文字列から```jsonと```を取り除く
        insights_json = insights_json.replace("```json", "").replace("```", "").strip()
        insights = json.loads(insights_json)

        # インサイトの可視化
        titles = [insight["title"] for insight in insights["insights"]]
        plt.figure(figsize=(10, 6))
        plt.barh(titles, [1, 2, 3])
        plt.xlabel('重要度')
        plt.title('データ分析から得られた主要インサイト')
        plt.tight_layout()
        plt.show()
    except:
        print("可視化に失敗しました。JSONの形式を確認してください。")

visualize_results(result["insights"], result["recommendations"])
````

リスト2.24 過去の会話をベクトル化して保存し、関連する会話を検索するためのモジュールをインポートする

````python
import os
from langchain.memory import VectorStoreRetrieverMemory
from langchain_community.vectorstores import FAISS
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain.chains import ConversationChain
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
load_dotenv()
import time

# Google APIキーの設定
os.environ["GOOGLE_API_KEY"] = os.getenv('GOOGLE_API_KEY')
````

リスト2.25 変数を定義して会話例を準備する

````python
# ベクトルストアの作成
embeddings = GoogleGenerativeAIEmbeddings(model="models/gemini-embedding-001")

# より多様な会話の例を準備
conversations = [
    {"input": "明日の東京の天気は？", "output": "明日の東京は晴れの予報です。最高気温は25度の見込みです。"},
    {"input": "大阪の週間天気予報を教えて", "output": "大阪の週間天気予報は、月曜から水曜は雨、木曜からは晴れの予報です。気温は22度から27度で推移する見込みです。"},
    {"input": "カレーの作り方を教えて", "output": "基本的なカレーの作り方は、玉ねぎと肉を炒め、カレールーを加えて煮込むだけです。野菜を加えるとより栄養価が高まります。"},
    {"input": "パスタのレシピはある？", "output": "シンプルなパスタのレシピなら、ニンニクとオリーブオイルで具材を炒め、茹でたパスタと和えるだけです。カルボナーラなら卵と生クリーム、チーズを混ぜると美味しいです。"},
    {"input": "東京ディズニーランドへの行き方", "output": "東京ディズニーランドへは、JR舞浜駅から徒歩で約10分です。東京駅からJR京葉線で舞浜駅まで約15分です。"},
    {"input": "京都の観光スポットおすすめは？", "output": "京都の人気観光スポットには、清水寺、伏見稲荷大社、金閣寺などがあります。季節によって桜や紅葉の名所も多いです。"},
    {"input": "英語の勉強法について", "output": "英語の効果的な勉強法は、毎日少しでも聞く練習をすること、簡単な本を読むこと、オンライン会話でアウトプットすることなどがあります。"},
    {"input": "プログラミング入門におすすめの言語は？", "output": "プログラミング入門ならPythonがおすすめです。文法がシンプルで、多くの学習リソースがあります。また、実用的なプログラムも比較的簡単に作れます。"}
]
````

リスト2.26 会話例をベクトル化

````python
# FAISベクトルストアの作成
texts = [f"ユーザー: {conv['input']}\nアシスタント: {conv['output']}" for conv in conversations]
vectorstore = FAISS.from_texts(texts, embeddings)
retriever = vectorstore.as_retriever(search_kwargs=dict(k=2))  # 上位2件の関連会話を取得

# ベクトルメモリの初期化
vector_memory = VectorStoreRetrieverMemory(retriever=retriever)

# 通常のLLMベースの会話を設定（比較用）
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0)
regular_chain = ConversationChain(
    llm=llm,
    verbose=True,
)

# ベクトルメモリを使用する会話を設定
vector_chain = ConversationChain(
    llm=llm,
    memory=vector_memory,
    verbose=True
)
````

リスト2.28 ベクトル検索のデモンストレーションの例

````python
# ベクトル検索をデモンストレーション
def demonstrate_vector_search():
    print("=" * 50)
    print("ベクトル検索のデモンストレーション")
    print("=" * 50)

    # いくつかのクエリで検索して結果を表示
    queries = [
        "天気について知りたい",
        "料理のレシピを教えて",
        "旅行先でおすすめの場所は？",
        "勉強方法について教えて"
    ]

    for query in queries:
        print(f"\nクエリ: {query}")
        result = vector_memory.load_memory_variables({"prompt": query})
        print(f"検索結果: {result['history']}")
        time.sleep(1)  # APIレート制限に配慮

demonstrate_vector_search()
````

リスト2.29 通常の会話とベクトルメモリを使用した会話の例

````python
# 通常の会話とベクトルメモリを使用した会話の比較
def compare_conversations():
    print("\n" + "=" * 50)
    print("通常の会話 vs ベクトルメモリを使用した会話")
    print("=" * 50)

    test_queries = [
        "明日の東京の天気はどうなる？",
        "美味しい食べ物の作り方は？",
        "東京への旅行を計画しています",
        "新しいスキルを身につけたい"
    ]

    for query in test_queries:
        print(f"\n質問: {query}")

        print("\n📝 通常の会話チェーン:")
        regular_response = regular_chain.predict(input=query)
        print(f"応答: {regular_response}")

        print("\n🔍 ベクトルメモリを使用した会話チェーン:")
        # まずメモリから関連情報を取得
        memory_result = vector_memory.load_memory_variables({"prompt": query})
        print(f"利用されたメモリ: {memory_result['history']}")

        # ベクトルメモリを使った応答を生成
        vector_response = vector_chain.predict(input=query)
        print(f"応答: {vector_response}")

        time.sleep(2)  # APIレート制限に配慮

compare_conversations()
````

リスト2.33 カスタマーサポートエージェントの例

````python
import os
from typing import List, Dict, Any
from datetime import datetime

from langchain.agents import initialize_agent, AgentType
from langchain.tools import BaseTool
from langchain_openai import ChatOpenAI
from langchain.memory import ConversationBufferMemory
from langchain.tools import tool
from langchain.pydantic_v1 import BaseModel, Field
from langchain.prompts import MessagesPlaceholder

# 環境変数の設定
os.environ['OPENAI_API_KEY'] = os.getenv('OPENAI_API_KEY')
os.environ['TAVILY_API_KEY'] = os.getenv('TAVILY_API_KEY')  # TavilyのAPIキーを設定

# モック製品データベース
PRODUCT_DB = {
    "P001": {"name": "スマートウォッチ Pro", "price": 25000, "stock": 15, "category": "電子機器"},
    "P002": {"name": "ワイヤレスイヤホン", "price": 15000, "stock": 8, "category": "電子機器"},
    "P003": {"name": "ポータブル充電器", "price": 5000, "stock": 20, "category": "電子機器"},
    "P004": {"name": "バックパック", "price": 8000, "stock": 12, "category": "バッグ"},
    "P005": {"name": "ヨガマット", "price": 3000, "stock": 25, "category": "スポーツ用品"}
}

# モック注文データベース
ORDER_DB = {
    "O001": {"customer_id": "C001", "products": ["P001"], "status": "配送中", "order_date": "2025-03-20", "delivery_date": "2025-03-25"},
    "O002": {"customer_id": "C002", "products": ["P002", "P003"], "status": "処理中", "order_date": "2025-03-22", "delivery_date": None},
    "O003": {"customer_id": "C001", "products": ["P004"], "status": "配達済み", "order_date": "2025-03-15", "delivery_date": "2025-03-18"},
    "O004": {"customer_id": "C003", "products": ["P005"], "status": "キャンセル", "order_date": "2025-03-10", "delivery_date": None}
}

# モック顧客データベース
CUSTOMER_DB = {
    "C001": {"name": "佐藤太郎", "email": "taro@example.com", "phone": "090-XXXX-XXXX", "address": "東京都新宿区"},
    "C002": {"name": "山田花子", "email": "hanako@example.com", "phone": "080-XXXX-XXXX", "address": "大阪府大阪市"},
    "C003": {"name": "鈴木一郎", "email": "ichiro@example.com", "phone": "070-XXXX-XXXX", "address": "北海道札幌市"}
}

# モックFAQデータベース
FAQ_DB = {
    "返品ポリシー": "商品到着後14日以内であれば、未使用・未開封の商品に限り返品可能です。返品送料はお客様負担となります。",
    "配送時間": "通常、ご注文確定後2〜5営業日以内に発送いたします。地域によって配送にかかる時間は異なります。",
    "支払い方法": "クレジットカード、コンビニ決済、銀行振込、代金引換に対応しています。",
    "会員登録": "公式ウェブサイトから無料で会員登録いただけます。会員特典として5%オフクーポンを進呈しています。",
    "製品保証": "電子機器は購入日から1年間の製品保証がついています。保証書と購入証明書を保管してください。"
}

# モックチケットシステム
TICKETS = {}

# ツール1: 製品検索ツール
class ProductQuerySchema(BaseModel):
    product_id: str = Field(description="検索する製品ID")

@tool("product_search", args_schema=ProductQuerySchema)
def search_product(product_id: str) -> str:
    """製品データベースから特定の製品情報を検索します。"""
    if product_id in PRODUCT_DB:
        product = PRODUCT_DB[product_id]
        return f"製品ID: {product_id}, 製品名: {product['name']}, 価格: {product['price']}円, 在庫: {product['stock']}個, カテゴリ: {product['category']}"
    return "指定された製品IDは見つかりませんでした。"

# ツール2: 注文ステータス確認ツール
class OrderQuerySchema(BaseModel):
    order_id: str = Field(description="確認する注文ID")

@tool("order_status", args_schema=OrderQuerySchema)
def check_order_status(order_id: str) -> str:
    """注文データベースから特定の注文のステータスを確認します。"""
    if order_id in ORDER_DB:
        order = ORDER_DB[order_id]
        products = [PRODUCT_DB[p]["name"] for p in order["products"]]
        delivery_info = f"配達予定日: {order['delivery_date']}" if order['delivery_date'] else "配達予定日: 未定"
        return f"注文ID: {order_id}, 状態: {order['status']}, 注文日: {order['order_date']}, {delivery_info}, 製品: {', '.join(products)}"
    return "指定された注文IDは見つかりませんでした。"

# ツール3: FAQ検索ツール
class FAQQuerySchema(BaseModel):
    keyword: str = Field(description="FAQのキーワード（例：返品、配送、支払い）")

@tool("faq_search", args_schema=FAQQuerySchema)
def search_faq(keyword: str) -> str:
    """FAQデータベースからキーワードに関連する情報を検索します。"""
    for topic, content in FAQ_DB.items():
        if keyword in topic:
            return f"【{topic}】\n{content}"
    return "指定されたキーワードに関するFAQは見つかりませんでした。"

# ツール4: チケット作成ツール
class TicketCreationSchema(BaseModel):
    customer_id: str = Field(description="顧客ID")
    issue: str = Field(description="問題の詳細")
    priority: str = Field(description="優先度（高/中/低）")

@tool("create_ticket", args_schema=TicketCreationSchema)
def create_support_ticket(customer_id: str, issue: str, priority: str) -> str:
    """カスタマーサポートチケットを作成します。"""
    if customer_id not in CUSTOMER_DB:
        return "指定された顧客IDは見つかりませんでした。"

    ticket_id = f"T{len(TICKETS) + 1:03d}"
    current_date = datetime.now().strftime("%Y-%m-%d")

    TICKETS[ticket_id] = {
        "customer_id": customer_id,
        "customer_name": CUSTOMER_DB[customer_id]["name"],
        "issue": issue,
        "priority": priority,
        "status": "新規",
        "created_at": current_date,
        "resolved_at": None
    }

    return f"チケットが作成されました。チケットID: {ticket_id}, 顧客: {CUSTOMER_DB[customer_id]['name']}, 優先度: {priority}, 作成日: {current_date}"

# ツール5: インターネット検索（Tavily API）
from langchain_community.tools.tavily_search import TavilySearchResults

search_tool = TavilySearchResults(max_results=3)

# すべてのツールをリストにまとめる
tools = [
    search_product,
    check_order_status,
    search_faq,
    create_support_ticket,
    search_tool
]

# 会話履歴を保持するメモリの初期化
memory = ConversationBufferMemory(memory_key="chat_history", return_messages=True)

# LLMの定義
llm = ChatOpenAI(
    model="gpt-4o-mini",  # 使用するモデル
    temperature=0.2,      # 少し創造性を持たせる
)

# システムメッセージの定義
system_message = """あなたは親切で有能なカスタマーサポートアシスタントです。
顧客の問い合わせに丁寧に対応し、必要に応じて適切なツールを使用して情報を提供してください。

以下のツールが利用可能です：
- product_search: 製品情報を検索
- order_status: 注文のステータスを確認
- faq_search: よくある質問の回答を検索
- create_ticket: サポートチケットを作成
- tavily_search_results_json: インターネット上の最新情報を検索

最初に顧客の問題を理解し、解決に最適なツールを選んでください。
複雑な問題の場合は、複数のツールを順番に使用することもできます。
解決できない問題については、サポートチケットを作成することを提案してください。

必ず日本語で応答し、敬語を使ってください。"""

# エージェントの初期化
agent = initialize_agent(
    tools,
    llm,
    agent=AgentType.OPENAI_FUNCTIONS,  # OpenAI Functionsを使用
    verbose=True,                      # 詳細な出力
    memory=memory,                     # 会話履歴
    agent_kwargs={
        "extra_prompt_messages": [
            MessagesPlaceholder(variable_name="chat_history")
        ],
        "system_message": system_message  # システムメッセージを直接設定
    }
)

# エージェントを実行する関数
def run_customer_support_agent(query: str) -> str:
    """カスタマーサポートエージェントを実行します。"""
    try:
        response = agent.run(query)
        return response
    except Exception as e:
        return f"エラーが発生しました: {str(e)}"

# 使用例
if __name__ == "__main__":
    # 製品情報の問い合わせ
    print("==== 製品情報の問い合わせ ====")
    print(run_customer_support_agent("スマートウォッチの情報を教えてください"))

    # 注文状況の確認
    print("\n==== 注文状況の確認 ====")
    print(run_customer_support_agent("注文番号O001の状況を確認したいです"))

    # 返品ポリシーについての質問
    print("\n==== 返品ポリシーについての質問 ====")
    print(run_customer_support_agent("返品の方法について教えてください"))

    # 複合的な問い合わせ
    print("\n==== 複合的な問い合わせ ====")
    print(run_customer_support_agent("先日購入したバックパックが破れていました。返品したいのですが、手続きを教えてください。"))
````
