リスト1.1　自律的なAIエージェントの意思決定プロセスの模擬コード

```python
class AutonomousAgent:
    def __init__(self, name, capabilities):
        self.name = name
        self.capabilities = capabilities
        self.known_situations = {}
        self.predefined_responses = {}

    def autonomous_decision(self, situation):
        # 既知の状況には事前に定義された応答を返す
        if situation in self.known_situations:
            return self.predefined_responses[situation]

        # 未知の状況に対しては独自に判断
        analysis = self._analyze_situation(situation)
        decision = self._make_decision(analysis)

        # 経験として記録
        self.known_situations[situation] = analysis
        self.predefined_responses[situation] = decision

        return decision

    def _analyze_situation(self, situation):
        # 状況を分析する内部プロセス
        # 実際の実装ではここに機械学習モデルや推論エンジンが入る
        return f"Analysis of {situation}"

    def _make_decision(self, analysis):
        # 分析に基づいて意思決定を行う
        return f"Decision based on {analysis}"
```

リスト1.2 意思決定のための選択肢評価システムの模擬コード

```python
def make_decision(self, observations, goal):
    # 現在の環境観測に基づいて可能な行動を列挙
    possible_actions = self._enumerate_actions(observations)

    # 各行動の結果を予測
    predicted_outcomes = {}
    for action in possible_actions:
        predicted_outcomes[action] = self._predict_outcome(observations, action)

    # 目標達成に最も寄与する行動を選択
    best_action = None
    best_score = float('-inf')

    for action, outcome in predicted_outcomes.items():
        score = self._evaluate_outcome(outcome, goal)
        if score > best_score:
            best_score = score
            best_action = action

    return best_action

```

リスト1.3 複雑な目標を段階的に実行する計画立案プロセスの模擬コード

```python
def plan_for_complex_goal(self, main_goal):
    # 目標を小さなサブタスクに分解
    sub_tasks = self._decompose_goal(main_goal)

    # 各サブタスクの依存関係を特定
    task_dependencies = self._identify_dependencies(sub_tasks)

    # タスクの実行順序を決定（トポロジカルソート）
    execution_order = self._topological_sort(sub_tasks, task_dependencies)

    # 実行計画を作成
    plan = []
    for task in execution_order:
        required_actions = self._plan_actions_for_task(task)
        plan.extend(required_actions)

    return plan
```

リスト1.4 マルチエージェント 協調システムの模擬コード

```python
def collaborate_with_other_agents(self, shared_goal, other_agents):
    # タスクをエージェント間で分担
    task_allocation = self._distribute_tasks(shared_goal, len(other_agents) + 1)

    # 自分のタスクを特定
    my_tasks = task_allocation[self.id]

    # 他のエージェントに割り当てられたタスクを通知
    for agent_id, tasks in task_allocation.items():
        if agent_id != self.id:
            agent = other_agents[agent_id]
            self._communicate(agent, {
                "message_type": "task_assignment",
                "tasks": tasks,
                "shared_goal": shared_goal
            })

    # 自分のタスクを実行
    my_results = self._execute_tasks(my_tasks)

    # 他のエージェントの結果を収集
    all_results = {self.id: my_results}
    for agent_id in task_allocation.keys():
        if agent_id != self.id:
            agent = other_agents[agent_id]
            agent_results = self._request_results(agent)
            all_results[agent_id] = agent_results

    # 結果を統合
    final_result = self._integrate_results(all_results, shared_goal)

    return final_result
```

リスト1.5 環境認識システムの基本構造を表す模擬コード

```python
def perceive_environment(self):
    # 各種センサーからデータを収集
    visual_data = self._collect_from_cameras()
    audio_data = self._collect_from_microphones()
    text_data = self._collect_from_user_input()
    api_data = self._collect_from_apis()

    # 収集したデータを前処理
    processed_visual = self._preprocess_visual(visual_data)
    processed_audio = self._preprocess_audio(audio_data)
    processed_text = self._preprocess_text(text_data)
    processed_api = self._preprocess_api(api_data)

    # 統合された環境認識を形成
    environment_state = {
        "visual": processed_visual,
        "audio": processed_audio,
        "text": processed_text,
        "api": processed_api,
        "timestamp": self._get_current_time()
    }

    # 環境状態をメモリに保存
    self.memory.add_perception(environment_state)

    return environment_state
```

リスト1.6 行動実行システムの模擬コード

```python
def execute_action(self, action_plan):
    # 行動プランを分解して実行可能なコマンドに変換
    commands = self._translate_to_commands(action_plan)

    # 各コマンドを順次実行
    results = []
    for cmd in commands:
        # コマンドのタイプに応じて適切なアクチュエータを選択
        if cmd["type"] == "movement":
            result = self._execute_movement(cmd["parameters"])
        elif cmd["type"] == "communication":
            result = self._execute_communication(cmd["parameters"])
        elif cmd["type"] == "data_processing":
            result = self._execute_processing(cmd["parameters"])
        else:
            result = {"status": "error", "message": f"Unknown command type: {cmd['type']}"}

        results.append(result)

        # 実行結果に問題があれば中断
        if result["status"] == "error":
            break

    # 行動の記録を保存
    self.memory.add_action_record({
        "plan": action_plan,
        "commands": commands,
        "results": results,
        "timestamp": self._get_current_time()
    })

    return results
```

リスト1.7 LLM駆動型AIエージェントの基本を表す模擬コード

```python
class LLMPoweredAgent:
    def __init__(self, llm_model, tools=None):
        self.llm = llm_model
        self.tools = tools or {}
        self.conversation_history = []

    def process_input(self, user_input):
        # 入力を会話履歴に追加
        self.conversation_history.append({"role": "user", "content": user_input})

        # LLMを使用して応答を生成
        prompt = self._prepare_prompt()
        llm_response = self.llm.generate(prompt)

        # ツール使用の指示があるか確認
        tool_calls = self._extract_tool_calls(llm_response)

        if tool_calls:
            # ツールを実行
            tool_results = self._execute_tools(tool_calls)

            # ツール結果を含めて再度応答を生成
            self.conversation_history.append({"role": "assistant", "content": llm_response})
            self.conversation_history.append({"role": "system", "content": f"Tool results: {tool_results}"})

            prompt = self._prepare_prompt()
            final_response = self.llm.generate(prompt)

            # 最終応答を履歴に追加
            self.conversation_history.append({"role": "assistant", "content": final_response})
            return final_response
        else:
            # ツール使用なしの場合は直接応答
            self.conversation_history.append({"role": "assistant", "content": llm_response})
            return llm_response
```

リスト1.8 外部ツール連携機能の模擬コード

```python
def use_external_tools(self, task_description):
    # タスクに必要なツールを特定
    required_tools = self._identify_required_tools(task_description)

    # 各ツールを実行して結果を取得
    tool_results = {}
    for tool_name in required_tools:
        if tool_name in self.available_tools:
            tool = self.available_tools[tool_name]

            # ツールの実行に必要なパラメータを準備
            parameters = self._prepare_tool_parameters(tool_name, task_description)

            # ツールを実行
            result = tool.execute(parameters)
            tool_results[tool_name] = result
        else:
            tool_results[tool_name] = {"error": f"Tool {tool_name} not available"}

    # ツールの実行結果を統合して最終レスポンスを生成
    integrated_response = self._integrate_tool_results(tool_results, task_description)

    return integrated_response
```

リスト1.11 2000年代強化学習エージェントの基本実装イメージ

```python
# 2000年代の強化学習エージェントの簡略化した例
import numpy as np

class RLAgent:
    def __init__(self, states, actions):
        self.q_table = np.zeros((states, actions))
        self.learning_rate = 0.1
        self.discount_factor = 0.9
        self.exploration_rate = 0.1

    def choose_action(self, state):
        # 探索と活用のバランス
        if np.random.random() < self.exploration_rate:
            return np.random.randint(0, self.q_table.shape[1])  # ランダム行動
        else:
            return np.argmax(self.q_table[state, :])  # 最良の行動

    def learn(self, state, action, reward, next_state):
        # Q学習によるアップデート
        best_next_action = np.argmax(self.q_table[next_state, :])
        td_target = reward + self.discount_factor * self.q_table[next_state, best_next_action]
        td_error = td_target - self.q_table[state, action]
        self.q_table[state, action] += self.learning_rate * td_error

```

リスト1.15 複数ツールを統合したAIエージェントの 実装例

```python
from langchain.agents import initialize_agent, Tool, AgentType
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()
import os

# APIキーの設定
os.environ['OPENAI_API_KEY'] = os.getenv('OPENAI_API_KEY')

# 実際の人口データ（仮の値、UNデータに基づく近似値）
WORLD_POPULATION_2000 = 6_127_700_000  # 2000年の世界人口（約61.27億人）
WORLD_POPULATION_2023 = 8_045_311_447  # 2023年の世界人口（約80.45億人）

# 計算機ツール：LLMが自然言語混じりの式を渡してもクラッシュせず、エラーを返して再試行させる
def calculate(expr: str) -> str:
    try:
        return str(eval(expr))
    except Exception as e:
        return f"計算エラー: 数値のみのPython数式を入力してください（例: (8045311447 - 6127700000) / 6127700000 * 100）。詳細: {e}"

# ツールの定義
tools = [
    Tool(
        name="検索",
        func=lambda q: f"検索結果 for {q}",  # 仮の実装
        description="最新の情報を検索するのに役立ちます"
    ),
    Tool(
        name="計算機",
        func=calculate,
        description="数学的計算を実行するのに役立ちます。入力は数値のみのPython数式にしてください（変数名や日本語を含めないこと）"
    ),
    Tool(
        name="データベース",
        func=lambda query: {
            "2023年の人口データと2000年の人口データ": f"{WORLD_POPULATION_2023}, {WORLD_POPULATION_2000}"
        }.get(query, f"データベース結果 for {query}"),
        description="構造化データにアクセスするのに役立ちます"
    )
]

# LLMのインスタンス化（gpt-3.5-turbo-instructは廃止されたため、チャットモデルを使用）
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

# エージェントの初期化
agent = initialize_agent(
    tools=tools,
    llm=llm,
    agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
    verbose=True
)

# エージェントの実行（非推奨の run の代わりに invoke を使用）
result = agent.invoke("2023年の世界人口と2000年の人口を調べて、成長率を計算してください")
print(result)
```

リスト1.16 LangChainを用いた基本的なAIエージェント実装例

```python
from langchain.agents import AgentType, initialize_agent, load_tools
from langchain.callbacks import StreamingStdOutCallbackHandler
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.tools import BaseTool, Tool
import os
import requests
from dotenv import load_dotenv
load_dotenv()
import ast
import time

# 環境変数の設定（GOOGLE_API_KEY を使用）
os.environ["GOOGLE_API_KEY"] = os.getenv('GOOGLE_API_KEY')

# Wikimedia APIの個人用アクセストークン（Wikipedia検索のレート制限緩和用。
# 未設定・無効な場合でも匿名アクセスにフォールバックする）
WIKIMEDIA_API_TOKEN = os.getenv("WIKIMEDIA_API_TOKEN")
WIKIPEDIA_USER_AGENT = "LangChainAIAgent-Book/1.0 (local study project)"

# カスタム Python REPL ツールの定義
class PythonREPLTool(BaseTool):
    name: str = "python_repl"
    description: str = "Pythonコードを実行し、その結果を返すツールです。"

    def _run(self, code: str) -> str:
        try:
            tree = ast.parse(code)
            exec_globals = {}
            exec(code, exec_globals)
            return str(exec_globals.get("result", "コードは正常に実行されました"))
        except Exception as e:
            return f"Error: {str(e)}"

    def _arun(self, code: str):
        raise NotImplementedError("非同期実行はサポートされていません")

def create_basic_agent():
    """シンプルなエージェントを作成する"""
    # LLM の初期化（Google Gemini 2.5 Flash を利用）
    llm = ChatGoogleGenerativeAI(
        model="gemini-2.5-flash",  # 必要に応じて"gemini-2.0-flash-thinking" 等に変更可能
        temperature=0,
        callbacks=[StreamingStdOutCallbackHandler()]
    )

    # ツールのロード（例として llm-math と wikipedia を利用）
    tools = load_tools(
        ["llm-math"],
        llm=llm
    )

    # Wikipedia検索：Wikimedia REST APIを直接呼び出す。WIKIMEDIA_API_TOKENが
    # 設定されていれば Authorization: Bearer ヘッダーを付け、匿名（10req/分）より
    # 大幅に高いレート制限（認証済みは数千req/時）で利用できる。トークンが無い・
    # 無効な場合も匿名アクセスにフォールバックし、失敗時は観測結果としてエラー
    # 文字列を返すことでエージェント全体のクラッシュを防ぐ。
    def _wikipedia_headers() -> dict:
        headers = {"User-Agent": WIKIPEDIA_USER_AGENT}
        if WIKIMEDIA_API_TOKEN:
            headers["Authorization"] = f"Bearer {WIKIMEDIA_API_TOKEN}"
        return headers

    def _wikipedia_search_once(query: str, top_k: int = 2) -> str:
        headers = _wikipedia_headers()
        search_resp = requests.get(
            "https://api.wikimedia.org/core/v1/wikipedia/en/search/page",
            params={"q": query, "limit": top_k},
            headers=headers,
            timeout=10,
        )
        search_resp.raise_for_status()
        pages = search_resp.json().get("pages", [])
        if not pages:
            return "該当するWikipediaの記事が見つかりませんでした。"

        summaries = []
        for page in pages:
            summary_resp = requests.get(
                f"https://en.wikipedia.org/api/rest_v1/page/summary/{page['key']}",
                headers=headers,
                timeout=10,
            )
            if summary_resp.ok:
                extract = summary_resp.json().get("extract", "")
                if extract:
                    summaries.append(f"Page: {page['title']}\nSummary: {extract}")
        return "\n\n".join(summaries) if summaries else "該当するWikipediaの記事が見つかりませんでした。"

    _last_wikipedia_call = {"t": 0.0}

    def safe_wikipedia_search(query: str) -> str:
        # トークンがあれば連続呼び出し間隔・再試行待機を短くできる
        min_interval = 1 if WIKIMEDIA_API_TOKEN else 5
        retry_wait = 10 if WIKIMEDIA_API_TOKEN else 30

        elapsed = time.time() - _last_wikipedia_call["t"]
        if elapsed < min_interval:
            time.sleep(min_interval - elapsed)

        for attempt in range(2):  # 初回 + 1回だけ再試行
            try:
                result = _wikipedia_search_once(query)
                _last_wikipedia_call["t"] = time.time()
                return result
            except Exception as e:
                _last_wikipedia_call["t"] = time.time()
                if attempt == 0:
                    time.sleep(retry_wait)
                    continue
                return f"Wikipedia検索が一時的に失敗しました（レート制限、またはWIKIMEDIA_API_TOKENの設定を確認してください）。詳細: {e}"

    tools.append(
        Tool(
            name="wikipedia",
            func=safe_wikipedia_search,
            description="Wikipediaで情報を検索します。検索したい用語や質問を入力してください。"
        )
    )
    tools.append(PythonREPLTool())  # カスタムツールの追加

    # エージェントの初期化
    agent = initialize_agent(
        tools,
        llm,
        agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION,
        handle_parsing_errors=True,
        verbose=True
    )
    return agent

def run_agent_demo():
    """エージェントのデモを実行する"""
    agent = create_basic_agent()
    # 複雑なタスクを実行する例
    agent.invoke({
        "input": "誰が2022年にノーベル物理学賞をとりましたか？ 研究をサマリして、簡単なPythonコードでその分野の基本原則を表して。日本語で考え日本語で返答して。"
    })

if __name__ == "__main__":
    run_agent_demo()
```
